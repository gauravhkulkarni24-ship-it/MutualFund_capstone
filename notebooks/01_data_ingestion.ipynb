```python
import pandas as pd
import numpy as np

RAW = "bluestock/data/raw"

files = {
    "01_fund_master"          : f"{RAW}/01_fund_master.csv",
    "02_nav_history"          : f"{RAW}/02_nav_history.csv",
    "03_aum_by_fund_house"    : f"{RAW}/03_aum_by_fund_house.csv",
    "04_monthly_sip_inflows"  : f"{RAW}/04_monthly_sip_inflows.csv",
    "05_category_inflows"     : f"{RAW}/05_category_inflows.csv",
    "06_industry_folio_count" : f"{RAW}/06_industry_folio_count.csv",
    "07_scheme_performance"   : f"{RAW}/07_scheme_performance.csv",
    "08_investor_transactions": f"{RAW}/08_investor_transactions.csv",
    "09_portfolio_holdings"   : f"{RAW}/09_portfolio_holdings.csv",
    "10_benchmark_indices"    : f"{RAW}/10_benchmark_indices.csv",
}

datasets = {}

for name, path in files.items():
    print(f"\n{'='*55}")
    print(f"📄 {name.upper()}")
    print(f"{'='*55}")
    df = pd.read_csv(path)
    datasets[name] = df
    print(f"Shape   : {df.shape}")
    print(f"Columns : {df.columns.tolist()}")
    print(f"Dtypes  :\n{df.dtypes}")
    print(f"\nHead(3) :\n{df.head(3)}")
    nulls = df.isnull().sum()
    if nulls.any():
        print(f"\n Nulls : {nulls[nulls>0].to_dict()}")
    else:
        print(f"\n No nulls")
    dups = df.duplicated().sum()
    print(f"Duplicates : {dups}")
```

```python
df = datasets["01_fund_master"]

print("="*55)
print("FUND MASTER EXPLORATION")
print("="*55)

print(f"\n Total Schemes : {len(df)}")

print(f"\n Fund Houses ({df['fund_house'].nunique()}):")
print(df['fund_house'].value_counts().to_string())

print(f"\n🗂 Categories:")
print(df['category'].value_counts().to_string())

print(f"\n Sub-Categories:")
print(df['sub_category'].value_counts().to_string())

print(f"\n Risk Categories:")
print(df['risk_category'].value_counts().to_string())

print(f"\n Plans:")
print(df['plan'].value_counts().to_string())

print(f"\n Expense Ratio:")
print(f"   Min : {df['expense_ratio_pct'].min()}")
print(f"   Max : {df['expense_ratio_pct'].max()}")
print(f"   Avg : {df['expense_ratio_pct'].mean():.3f}")
```

```python
import json
from datetime import datetime

fund_master = datasets["01_fund_master"]
nav_history = datasets["02_nav_history"]

master_codes = set(fund_master["amfi_code"].astype(str))
nav_codes    = set(nav_history["amfi_code"].astype(str))

matched = master_codes & nav_codes
missing = master_codes - nav_codes

print("="*55)
print("AMFI CODE VALIDATION")
print("="*55)
print(f"\n Master codes  : {len(master_codes)}")
print(f" NAV codes     : {len(nav_codes)}")
print(f" Matched       : {len(matched)}")
print(f"  Missing       : {len(missing)}")
coverage = len(matched)/len(master_codes)*100
print(f" Coverage      : {coverage:.1f}%")
verdict = "PASS " if coverage == 100 else "WARN "
print(f" Verdict       : {verdict}")

# Save quality report
report = {
    "date"          : datetime.now().strftime("%Y-%m-%d %H:%M"),
    "master_codes"  : len(master_codes),
    "nav_codes"     : len(nav_codes),
    "matched"       : len(matched),
    "missing"       : sorted(missing),
    "coverage_pct"  : round(coverage, 2),
    "verdict"       : verdict
}
os.makedirs("bluestock/reports", exist_ok=True)
with open("bluestock/reports/data_quality_day1.json", "w") as f:
    json.dump(report, f, indent=2)
print("\n Report saved → bluestock/reports/data_quality_day1.json")
```

```python
import requests

SCHEMES = {
    125497: "HDFC_Top100_Direct",
    119551: "SBI_Bluechip",
    120503: "ICICI_Bluechip",
    118632: "Nippon_LargeCap",
    119092: "Axis_Bluechip",
    120841: "Kotak_Bluechip",
}

all_nav = []

for code, name in SCHEMES.items():
    print(f"Fetching {name} ({code})...", end=" ")
    try:
        r    = requests.get(f"https://api.mfapi.in/mf/{code}", timeout=15)
        data = r.json()
        df   = pd.DataFrame(data["data"])
        df["amfi_code"]   = code
        df["scheme_name"] = data["meta"]["scheme_name"]
        df["fund_house"]  = data["meta"]["fund_house"]
        df.to_csv(f"bluestock/data/raw/nav_live_{code}.csv", index=False)
        all_nav.append(df)
        print(f"✅ {len(df):,} records")
    except Exception as e:
        print(f"❌ {e}")

if all_nav:
    combined = pd.concat(all_nav, ignore_index=True)
    combined.to_csv("bluestock/data/raw/nav_live_combined.csv", index=False)
    print(f"\n✅ Combined saved → {len(combined):,} total rows")
```

```python
PROC = "bluestock/data/processed"

# 1. Fund Master
df = datasets["01_fund_master"].copy()
df["launch_date"]       = pd.to_datetime(df["launch_date"], errors="coerce")
df["expense_ratio_pct"] = pd.to_numeric(df["expense_ratio_pct"], errors="coerce")
df.drop_duplicates(subset=["amfi_code"], inplace=True)
df.to_csv(f"{PROC}/01_fund_master_clean.csv", index=False)
print(f" Fund Master       : {len(df):>8,} rows")

# 2. NAV History
df = datasets["02_nav_history"].copy()
df["date"] = pd.to_datetime(df["date"], errors="coerce")
df["nav"]  = pd.to_numeric(df["nav"],  errors="coerce")
df.dropna(subset=["date","nav"], inplace=True)
df = df[df["nav"] > 0]
df.drop_duplicates(subset=["amfi_code","date"], inplace=True)
df.sort_values(["amfi_code","date"], inplace=True)
df["daily_return_pct"] = df.groupby("amfi_code")["nav"].pct_change() * 100
df.to_csv(f"{PROC}/02_nav_history_clean.csv", index=False)
print(f" NAV History       : {len(df):>8,} rows")

# 3. AUM
df = datasets["03_aum_by_fund_house"].copy()
df["date"] = pd.to_datetime(df["date"], errors="coerce")
df.dropna(inplace=True)
df.to_csv(f"{PROC}/03_aum_clean.csv", index=False)
print(f" AUM               : {len(df):>8,} rows")

# 4. SIP
df = datasets["04_monthly_sip_inflows"].copy()
df["month"]          = pd.to_datetime(df["month"], errors="coerce")
df["yoy_growth_pct"] = df["yoy_growth_pct"].fillna(0)
df.dropna(subset=["month","sip_inflow_crore"], inplace=True)
df.to_csv(f"{PROC}/04_sip_clean.csv", index=False)
print(f" SIP Inflows       : {len(df):>8,} rows")

# 5. Category Inflows
df = datasets["05_category_inflows"].copy()
df["month"] = pd.to_datetime(df["month"], errors="coerce")
df.dropna(inplace=True)
df.to_csv(f"{PROC}/05_category_clean.csv", index=False)
print(f" Category Inflows  : {len(df):>8,} rows")

# 6. Folio Count
df = datasets["06_industry_folio_count"].copy()
df["month"] = pd.to_datetime(df["month"], errors="coerce")
df.dropna(inplace=True)
df.to_csv(f"{PROC}/06_folio_clean.csv", index=False)
print(f" Folio Count       : {len(df):>8,} rows")

# 7. Scheme Performance
df = datasets["07_scheme_performance"].copy()
for col in ["return_1yr_pct","return_3yr_pct","return_5yr_pct",
            "sharpe_ratio","sortino_ratio","beta","alpha",
            "max_drawdown_pct","std_dev_ann_pct"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")
df.to_csv(f"{PROC}/07_performance_clean.csv", index=False)
print(f" Scheme Performance: {len(df):>8,} rows")
# 8. Investor Transactions
df = datasets["08_investor_transactions"].copy()
df["transaction_date"]   = pd.to_datetime(df["transaction_date"], errors="coerce")
df["amount_inr"]         = pd.to_numeric(df["amount_inr"],        errors="coerce")
df["transaction_type"]   = df["transaction_type"].str.strip().str.title()
df = df[df["amount_inr"] > 0]
df.dropna(subset=["investor_id","amfi_code","transaction_date"], inplace=True)
df.to_csv(f"{PROC}/08_transactions_clean.csv", index=False)
print(f" Transactions      : {len(df):>8,} rows")

# 9. Portfolio Holdings
df = datasets["09_portfolio_holdings"].copy()
df["weight_pct"]      = pd.to_numeric(df["weight_pct"],      errors="coerce")
df["market_value_cr"] = pd.to_numeric(df["market_value_cr"], errors="coerce")
df.dropna(inplace=True)
df.to_csv(f"{PROC}/09_holdings_clean.csv", index=False)
print(f" Holdings          : {len(df):>8,} rows")

# 10. Benchmark
df = datasets["10_benchmark_indices"].copy()
df["date"]        = pd.to_datetime(df["date"],        errors="coerce")
df["close_value"] = pd.to_numeric(df["close_value"],  errors="coerce")
df.dropna(inplace=True)
df.sort_values(["index_name","date"], inplace=True)
df.to_csv(f"{PROC}/10_benchmark_clean.csv", index=False)
print(f"Benchmark         : {len(df):>8,} rows")

print("\n All 10 files cleaned and saved!")
```

```python
from sqlalchemy import create_engine
import sqlite3

engine = create_engine("sqlite:///bluestock/data/db/bluestock_mf.db")
PROC   = "bluestock/data/processed"

# Load cleaned files
df_fund  = pd.read_csv(f"{PROC}/01_fund_master_clean.csv")
df_nav   = pd.read_csv(f"{PROC}/02_nav_history_clean.csv")
df_aum   = pd.read_csv(f"{PROC}/03_aum_clean.csv")
df_sip   = pd.read_csv(f"{PROC}/04_sip_clean.csv")
df_cat   = pd.read_csv(f"{PROC}/05_category_clean.csv")
df_fol   = pd.read_csv(f"{PROC}/06_folio_clean.csv")
df_perf  = pd.read_csv(f"{PROC}/07_performance_clean.csv")
df_txn   = pd.read_csv(f"{PROC}/08_transactions_clean.csv")
df_hold  = pd.read_csv(f"{PROC}/09_holdings_clean.csv")
df_bm    = pd.read_csv(f"{PROC}/10_benchmark_clean.csv")

# Load all tables
df_fund.to_sql("dim_fund",              engine, if_exists="replace", index=False)
print(f"✅ dim_fund              : {len(df_fund):>8,} rows")

df_txn[["investor_id","state","city","city_tier","age_group",
        "gender","annual_income_lakh"]].drop_duplicates(
    subset=["investor_id"]).to_sql(
    "dim_investor", engine, if_exists="replace", index=False)
print(f"✅ dim_investor          : {df_txn['investor_id'].nunique():>8,} rows")

df_nav.to_sql("fact_nav",               engine, if_exists="replace", index=False)
print(f"✅ fact_nav              : {len(df_nav):>8,} rows")

df_txn.to_sql("fact_transactions",      engine, if_exists="replace", index=False)
print(f"✅ fact_transactions     : {len(df_txn):>8,} rows")

df_perf.to_sql("fact_performance",      engine, if_exists="replace", index=False)
print(f"✅ fact_performance      : {len(df_perf):>8,} rows")

df_aum.to_sql("fact_aum",               engine, if_exists="replace", index=False)
print(f"✅ fact_aum              : {len(df_aum):>8,} rows")

df_sip.to_sql("fact_sip",               engine, if_exists="replace", index=False)
print(f"✅ fact_sip              : {len(df_sip):>8,} rows")

df_hold.to_sql("fact_holdings",         engine, if_exists="replace", index=False)
print(f"✅ fact_holdings         : {len(df_hold):>8,} rows")

df_bm.to_sql("fact_benchmark",          engine, if_exists="replace", index=False)
print(f"✅ fact_benchmark        : {len(df_bm):>8,} rows")

df_cat.to_sql("fact_category_inflows",  engine, if_exists="replace", index=False)
print(f"✅ fact_category_inflows : {len(df_cat):>8,} rows")

df_fol.to_sql("fact_folio",             engine, if_exists="replace", index=False)
print(f"✅ fact_folio            : {len(df_fol):>8,} rows")

print("\n✅ Database ready → bluestock/data/db/bluestock_mf.db")
```

```python
conn = sqlite3.connect("bluestock/data/db/bluestock_mf.db")

tables = ["dim_fund","dim_investor","fact_nav","fact_transactions",
          "fact_performance","fact_aum","fact_sip","fact_holdings",
          "fact_benchmark","fact_category_inflows","fact_folio"]

print("="*55)
print("DATABASE VALIDATION")
print("="*55)
for t in tables:
    cnt = pd.read_sql(f"SELECT COUNT(*) AS c FROM {t}", conn).iloc[0,0]
    print(f" {t:<30} : {cnt:>8,} rows")

# Quick test query
print("\n Top 5 Funds by AUM:")
print(pd.read_sql("""
    SELECT scheme_name, fund_house, aum_crore
    FROM fact_performance
    ORDER BY aum_crore DESC LIMIT 5
""", conn).to_string(index=False))

conn.close()
print("\n All tables validated!")
```
