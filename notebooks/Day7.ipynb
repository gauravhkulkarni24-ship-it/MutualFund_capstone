```python
!pip install python-pptx -q

from pptx import Presentation
from pptx.util import Inches, Pt, Emu
from pptx.dml.color import RGBColor
from pptx.enum.text import PP_ALIGN
import datetime

prs = Presentation()
prs.slide_width  = Inches(13.33)
prs.slide_height = Inches(7.5)

# Colors
DARK_BLUE  = RGBColor(0x0A, 0x16, 0x28)
BLUE       = RGBColor(0x34, 0x98, 0xDB)
GREEN      = RGBColor(0x2E, 0xCC, 0x71)
WHITE      = RGBColor(0xFF, 0xFF, 0xFF)
LIGHT_GRAY = RGBColor(0x95, 0xA5, 0xA6)
ORANGE     = RGBColor(0xF3, 0x9C, 0x12)


def add_slide(prs, layout_idx=6):
    layout = prs.slide_layouts[layout_idx]
    return prs.slides.add_slide(layout)


def set_bg(slide, color):
    background = slide.background
    fill       = background.fill
    fill.solid()
    fill.fore_color.rgb = color


def add_text(slide, text, left, top, width, height,
             fontsize=24, bold=False, color=WHITE,
             align=PP_ALIGN.LEFT):
    txBox = slide.shapes.add_textbox(
        Inches(left), Inches(top),
        Inches(width), Inches(height))
    tf   = txBox.text_frame
    tf.word_wrap = True
    p    = tf.paragraphs[0]
    p.alignment = align
    run  = p.add_run()
    run.text = text
    run.font.size  = Pt(fontsize)
    run.font.bold  = bold
    run.font.color.rgb = color
    return txBox


# ── SLIDE 1: Title ────────────────────────────────────────
slide = add_slide(prs)
set_bg(slide, DARK_BLUE)
add_text(slide, "Bluestock Fintech Pvt. Ltd.",
         1, 0.8, 11, 1, 18, False, BLUE, PP_ALIGN.CENTER)
add_text(slide, "Mutual Fund Analytics Platform",
         1, 1.6, 11, 1.5, 32, True, WHITE, PP_ALIGN.CENTER)
add_text(slide, "Capstone Project  ·  7 Days  ·  June 2026",
         1, 3.0, 11, 0.8, 16, False, LIGHT_GRAY, PP_ALIGN.CENTER)
add_text(slide, "Python  ·  SQL  ·  Power BI  ·  AMFI India  ·  mfapi.in",
         1, 3.8, 11, 0.8, 14, False, BLUE, PP_ALIGN.CENTER)

stats_text = "40 Schemes    |    46,000+ NAV Records    |    32,778 Transactions    |    10 Fund Houses"
add_text(slide, stats_text,
         1, 5.5, 11, 0.8, 13, False, GREEN, PP_ALIGN.CENTER)

# ── SLIDE 2: Problem Statement ────────────────────────────
slide = add_slide(prs)
set_bg(slide, DARK_BLUE)
add_text(slide, "Problem Statement", 0.5, 0.3, 12, 1,
         28, True, WHITE, PP_ALIGN.LEFT)
problems = [
    "P1: NAV, AUM, SIP data scattered across multiple AMFI sources",
    "P2: No unified platform to compare funds on risk-adjusted basis",
    "P3: Investors cannot track benchmark outperformance easily",
    "P4: AMCs lack visibility into demographic investment patterns",
    "P5: Monthly MF reports are static PDFs — no drill-down capability",
]
for i, p in enumerate(problems):
    add_text(slide, p, 0.5, 1.3 + i*1.0, 12, 0.8,
             14, False, LIGHT_GRAY)

# ── SLIDE 3: Objectives ───────────────────────────────────
slide = add_slide(prs)
set_bg(slide, DARK_BLUE)
add_text(slide, "Project Objectives", 0.5, 0.3, 12, 1,
         28, True, WHITE)
objectives = [
    "O1  Build automated ETL pipeline from AMFI + mfapi.in",
    "O2  Design normalised SQLite star schema (5+ tables)",
    "O3  Perform EDA with 16+ charts on NAV, AUM, SIP data",
    "O4  Compute Sharpe, Sortino, Alpha, Beta, VaR, Max Drawdown",
    "O5  Build 4-page interactive Power BI dashboard",
    "O6  Analyse investor demographics and transaction patterns",
    "O7  Benchmark funds vs Nifty 50 / Nifty 100",
    "O8  Document findings in PDF report + 12-slide deck",
]
for i, o in enumerate(objectives):
    add_text(slide, o, 0.5, 1.2 + i*0.72, 12, 0.6,
             13, False, GREEN if i % 2 == 0 else LIGHT_GRAY)

# ── SLIDE 4: Data Sources ─────────────────────────────────
slide = add_slide(prs)
set_bg(slide, DARK_BLUE)
add_text(slide, "Data Sources & Datasets", 0.5, 0.3, 12, 1,
         28, True, WHITE)
datasets = [
    ("01_fund_master.csv",           "40 rows",    "40 real AMFI schemes"),
    ("02_nav_history.csv",           "46,000 rows","Daily NAV 2022–2026"),
    ("08_investor_transactions.csv", "32,778 rows","5,000 investors, 12 states"),
    ("10_benchmark_indices.csv",     "8,050 rows", "Nifty50, Nifty100, BSE"),
    ("04_monthly_sip_inflows.csv",   "48 rows",    "Real AMFI SIP data"),
    ("03_aum_by_fund_house.csv",     "90 rows",    "Quarterly AUM 2022–2025"),
]
for i, (f, rows, desc) in enumerate(datasets):
    y = 1.3 + i * 0.85
    add_text(slide, f,    0.5, y, 5.0, 0.6, 12, True,  BLUE)
    add_text(slide, rows, 5.5, y, 2.5, 0.6, 12, False, GREEN)
    add_text(slide, desc, 8.0, y, 5.0, 0.6, 12, False, LIGHT_GRAY)

# ── SLIDE 5: Architecture ─────────────────────────────────
slide = add_slide(prs)
set_bg(slide, DARK_BLUE)
add_text(slide, "System Architecture — ETL Pipeline", 0.5, 0.3,
         12, 1, 28, True, WHITE)
layers = [
    ("EXTRACT",   "AMFI India + mfapi.in REST API + Local CSVs"),
    ("TRANSFORM", "Pandas cleaning · Type casting · Forward-fill · Validation"),
    ("LOAD",      "SQLite Star Schema · 12 tables · Indexed on amfi_code + date"),
    ("ANALYSE",   "Sharpe · Sortino · Alpha · Beta · VaR · Drawdown · Scorecard"),
    ("VISUALISE", "Power BI Dashboard · 4 pages · Slicers · KPI cards"),
]
colors_layer = [BLUE, GREEN, ORANGE, BLUE, GREEN]
for i, ((layer, desc), col) in enumerate(zip(layers, colors_layer)):
    y = 1.4 + i * 1.0
    add_text(slide, layer, 0.5, y, 2.5, 0.7, 14, True,  col)
    add_text(slide, f"→  {desc}", 3.0, y, 10, 0.7, 13, False, LIGHT_GRAY)

# ── SLIDE 6: EDA Highlights ───────────────────────────────
slide = add_slide(prs)
set_bg(slide, DARK_BLUE)
add_text(slide, "EDA Highlights", 0.5, 0.3, 12, 1, 28, True, WHITE)
findings = [
    "📈 SIP inflows grew from ₹11,517 Cr (Jan 2022) → ₹31,002 Cr (Dec 2025)",
    "🏛  SBI MF dominates with ₹12.5 Lakh Crore AUM (Dec 2025)",
    "📊 Industry folios doubled: 13.26 Cr → 26.12 Cr in 4 years",
    "🔗 High correlation (>0.85) among Large Cap fund NAV returns",
    "🏢 Banking & IT sectors dominate portfolio holdings (45%+ weight)",
    "👤 26-35 age group: highest SIP count | 46-55: highest lumpsum",
    "🗺  Maharashtra & Karnataka lead in total investment volume",
]
for i, f in enumerate(findings):
    add_text(slide, f, 0.5, 1.2 + i*0.82, 12, 0.7,
             13, False, LIGHT_GRAY)

# ── SLIDE 7: Performance Metrics ─────────────────────────
slide = add_slide(prs)
set_bg(slide, DARK_BLUE)
add_text(slide, "Fund Performance Metrics", 0.5, 0.3, 12, 1,
         28, True, WHITE)

top3 = df_score.head(3)
add_text(slide, "🏆 TOP 3 FUNDS BY COMPOSITE SCORE",
         0.5, 1.2, 12, 0.6, 14, True, GREEN)

for i, (_, row) in enumerate(top3.iterrows()):
    y = 1.9 + i * 1.2
    add_text(slide,
             f"#{i+1}  {row['scheme_name'][:45]}",
             0.5, y, 10, 0.6, 13, True, WHITE)
    add_text(slide,
             f"Score: {row['composite_score']:.1f}  |  "
             f"3Y CAGR: {row['cagr_3y_pct']:.2f}%  |  "
             f"Sharpe: {row['sharpe_ratio']:.3f}",
             0.5, y+0.55, 10, 0.5, 11, False, BLUE)

add_text(slide, "Risk Metrics Summary",
         0.5, 5.0, 12, 0.5, 13, True, ORANGE)
add_text(slide,
         f"Avg Sharpe: {df_master['sharpe_ratio'].mean():.2f}  |  "
         f"Best 3Y Return: {df_master['cagr_3y_pct'].max():.1f}%  |  "
         f"Avg Max Drawdown: {df_master['max_drawdown_pct'].mean():.1f}%",
         0.5, 5.6, 12, 0.6, 12, False, LIGHT_GRAY)

# ── SLIDE 8: Risk Analysis ────────────────────────────────
slide = add_slide(prs)
set_bg(slide, DARK_BLUE)
add_text(slide, "Risk Analysis — VaR & Drawdown", 0.5, 0.3,
         12, 1, 28, True, WHITE)
worst_var = df_var.nsmallest(3, "var_95_pct")
best_var  = df_var.nlargest(3,  "var_95_pct")

add_text(slide, "⚠️  Highest Risk (Worst VaR 95%)",
         0.5, 1.2, 6, 0.6, 14, True, RGBColor(0xE7, 0x4C, 0x3C))
for i, (_, row) in enumerate(worst_var.iterrows()):
    add_text(slide,
             f"  {row['scheme_name'][:35]}  →  VaR: {row['var_95_pct']:.3f}%",
             0.5, 1.9+i*0.7, 6, 0.5, 12, False, LIGHT_GRAY)

add_text(slide, "✅  Lowest Risk (Best VaR 95%)",
         7.0, 1.2, 6, 0.6, 14, True, GREEN)
for i, (_, row) in enumerate(best_var.iterrows()):
    add_text(slide,
             f"  {row['scheme_name'][:35]}  →  VaR: {row['var_95_pct']:.3f}%",
             7.0, 1.9+i*0.7, 6, 0.5, 12, False, LIGHT_GRAY)

add_text(slide,
         f"Avg VaR 95%: {df_var['var_95_pct'].mean():.3f}% per day  |  "
         f"Avg CVaR: {df_var['cvar_95_pct'].mean():.3f}% per day",
         0.5, 5.5, 12, 0.6, 13, False, BLUE)

# ── SLIDE 9: Dashboard Screenshots ───────────────────────
slide = add_slide(prs)
set_bg(slide, DARK_BLUE)
add_text(slide, "Power BI Dashboard — 4 Pages", 0.5, 0.3,
         12, 1, 28, True, WHITE)
pages = [
    ("Page 1", "Industry Overview",    "AUM, SIP Inflow, Folio KPIs"),
    ("Page 2", "Fund Performance",     "Risk-Return scatter, Sharpe ranking"),
    ("Page 3", "Investor Analytics",   "State, Age, T30/B30 distribution"),
    ("Page 4", "SIP & Market Trends",  "SIP timeline, Benchmark comparison"),
]
for i, (pg, title, desc) in enumerate(pages):
    y = 1.4 + i * 1.3
    add_text(slide, pg,    0.5, y, 1.5, 0.6, 14, True,  BLUE)
    add_text(slide, title, 2.2, y, 4.5, 0.6, 14, True,  WHITE)
    add_text(slide, desc,  2.2, y+0.55, 10, 0.5, 12, False, LIGHT_GRAY)

# ── SLIDE 10: Advanced Analytics ─────────────────────────
slide = add_slide(prs)
set_bg(slide, DARK_BLUE)
add_text(slide, "Advanced Analytics — Day 6", 0.5, 0.3,
         12, 1, 28, True, WHITE)
advanced = [
    ("Rolling Sharpe",    "90-day rolling risk-adjusted return for top 5 funds"),
    ("Cohort Analysis",   "Investors grouped by first transaction year"),
    ("SIP Continuity",    "Flagged at-risk investors with gap > 35 days"),
    ("Fund Recommender",  "Rule-based system matching funds to risk appetite"),
    ("Sector HHI",        "Herfindahl index measuring portfolio concentration"),
]
for i, (title, desc) in enumerate(advanced):
    y = 1.3 + i * 1.0
    add_text(slide, f"▶  {title}", 0.5, y, 4, 0.6, 14, True,  GREEN)
    add_text(slide, desc,          5.0, y, 8, 0.6, 13, False, LIGHT_GRAY)

# ── SLIDE 11: Key Findings ────────────────────────────────
slide = add_slide(prs)
set_bg(slide, DARK_BLUE)
add_text(slide, "Key Findings & Insights", 0.5, 0.3,
         12, 1, 28, True, WHITE)
key_findings = [
    "₹31,002 Cr SIP milestone — highest ever monthly inflow (Dec 2025)",
    "SBI MF is India's largest AMC at ₹12.5 Lakh Crore AUM",
    "Folio count doubled in 4 years — 97% growth (13.26 → 26.12 Cr)",
    "Mid-cap funds deliver best alpha; large-cap most stable",
    "26-35 age group drives SIP growth — 31% of all investors",
    "B30 cities growing 2.5x faster than T30 in transaction volume",
    "Funds with Sharpe > 1.0 consistently outperform on risk basis",
    "SIP continuity analysis shows 23% investors at-risk of lapsing",
]
for i, f in enumerate(key_findings):
    color = GREEN if i % 2 == 0 else LIGHT_GRAY
    add_text(slide, f"{'✅' if i%2==0 else '📌'}  {f}",
             0.5, 1.2 + i*0.75, 12, 0.6, 12, False, color)

# ── SLIDE 12: Thank You ───────────────────────────────────
slide = add_slide(prs)
set_bg(slide, DARK_BLUE)
add_text(slide, "Thank You", 1, 1.5, 11, 2,
         48, True, WHITE, PP_ALIGN.CENTER)
add_text(slide, "Bluestock Fintech Pvt. Ltd. — Capstone Project",
         1, 3.2, 11, 1, 18, False, BLUE, PP_ALIGN.CENTER)
add_text(slide, "Python  ·  SQL  ·  Power BI  ·  AMFI India",
         1, 4.0, 11, 1, 14, False, LIGHT_GRAY, PP_ALIGN.CENTER)
add_text(slide,
         "github.com/gauravhkulkarni24-ship-it/mf_analytics",
         1, 5.0, 11, 1, 13, False, GREEN, PP_ALIGN.CENTER)
add_text(slide, "June 2026",
         1, 5.8, 11, 0.8, 12, False, LIGHT_GRAY, PP_ALIGN.CENTER)

# Save
pptx_path = "bluestock/reports/Presentation.pptx"
prs.save(pptx_path)
print(f"Presentation saved → {pptx_path}")
print(f"   Slides : 12")
```

```python
import os

# ── Fix 1: Copy scripts to correct location ───────────────
os.makedirs("bluestock/scripts", exist_ok=True)
os.makedirs("bluestock/sql", exist_ok=True)

# data_ingestion.py
data_ingestion = """
import pandas as pd
import numpy as np
import os
import json
from datetime import datetime

RAW_DIR = 'data/raw'
CSV_FILES = {
    'fund_master'          : f'{RAW_DIR}/01_fund_master.csv',
    'nav_history'          : f'{RAW_DIR}/02_nav_history.csv',
    'aum_by_fund_house'    : f'{RAW_DIR}/03_aum_by_fund_house.csv',
    'monthly_sip_inflows'  : f'{RAW_DIR}/04_monthly_sip_inflows.csv',
    'category_inflows'     : f'{RAW_DIR}/05_category_inflows.csv',
    'industry_folio_count' : f'{RAW_DIR}/06_industry_folio_count.csv',
    'scheme_performance'   : f'{RAW_DIR}/07_scheme_performance.csv',
    'investor_transactions': f'{RAW_DIR}/08_investor_transactions.csv',
    'portfolio_holdings'   : f'{RAW_DIR}/09_portfolio_holdings.csv',
    'benchmark_indices'    : f'{RAW_DIR}/10_benchmark_indices.csv',
}

def load_and_inspect(csv_map):
    datasets  = {}
    anomalies = []
    print('=' * 60)
    print('DAY 1 - LOADING ALL 10 DATASETS')
    print('=' * 60)
    for name, path in csv_map.items():
        if not os.path.exists(path):
            print(f'MISSING: {path}')
            continue
        df = pd.read_csv(path)
        datasets[name] = df
        print(f'\\n{name.upper()}')
        print(f'  Shape   : {df.shape}')
        print(f'  Columns : {df.columns.tolist()}')
        print(f'  Head    :\\n{df.head(3)}')
        nulls = df.isnull().sum()
        if nulls.any():
            anomalies.append(f'{name}: nulls in {nulls[nulls>0].index.tolist()}')
    return datasets, anomalies

def explore_fund_master(df):
    print('\\nFUND MASTER EXPLORATION')
    print(f'Total Schemes : {len(df)}')
    print(f'Fund Houses   : {df[\"fund_house\"].nunique()}')
    print(df['fund_house'].value_counts().to_string())
    print(df['category'].value_counts().to_string())
    print(df['risk_category'].value_counts().to_string())

def validate_amfi_codes(fund_master, nav_history):
    master_codes = set(fund_master['amfi_code'].astype(str))
    nav_codes    = set(nav_history['amfi_code'].astype(str))
    matched      = master_codes & nav_codes
    coverage     = len(matched) / len(master_codes) * 100
    print(f'\\nAMFI CODE VALIDATION')
    print(f'Master codes : {len(master_codes)}')
    print(f'NAV codes    : {len(nav_codes)}')
    print(f'Matched      : {len(matched)}')
    print(f'Coverage     : {coverage:.1f}%')
    print(f'Verdict      : PASS' if coverage == 100 else f'Verdict : WARN')

if __name__ == '__main__':
    datasets, anomalies = load_and_inspect(CSV_FILES)
    if 'fund_master' in datasets:
        explore_fund_master(datasets['fund_master'])
    if 'fund_master' in datasets and 'nav_history' in datasets:
        validate_amfi_codes(datasets['fund_master'], datasets['nav_history'])
    print('\\nAnomalies:', anomalies)
    print('\\nDay 1 complete!')
"""

with open("bluestock/scripts/data_ingestion.py", "w") as f:
    f.write(data_ingestion)
print("✅ data_ingestion.py saved")

# ── etl_pipeline.py ───────────────────────────────────────
etl_pipeline = """
import pandas as pd
import numpy as np
import sqlite3
import os
from sqlalchemy import create_engine

RAW_DIR       = 'data/raw'
PROCESSED_DIR = 'data/processed'
DB_PATH       = 'data/db/bluestock_mf.db'

os.makedirs(PROCESSED_DIR, exist_ok=True)
os.makedirs('data/db', exist_ok=True)

engine = create_engine(f'sqlite:///{DB_PATH}')

def clean_and_load():
    print('DAY 2 - ETL PIPELINE')

    # Fund Master
    df = pd.read_csv(f'{RAW_DIR}/01_fund_master.csv')
    df['launch_date'] = pd.to_datetime(df['launch_date'], errors='coerce')
    df.to_csv(f'{PROCESSED_DIR}/01_fund_master_clean.csv', index=False)
    df.to_sql('dim_fund', engine, if_exists='replace', index=False)
    print(f'dim_fund : {len(df)} rows')

    # NAV History
    df = pd.read_csv(f'{RAW_DIR}/02_nav_history.csv')
    df['date'] = pd.to_datetime(df['date'], errors='coerce')
    df['nav']  = pd.to_numeric(df['nav'], errors='coerce')
    df.dropna(subset=['date','nav'], inplace=True)
    df = df[df['nav'] > 0]
    df['daily_return_pct'] = df.groupby('amfi_code')['nav'].pct_change() * 100
    df.to_csv(f'{PROCESSED_DIR}/02_nav_history_clean.csv', index=False)
    df.to_sql('fact_nav', engine, if_exists='replace', index=False)
    print(f'fact_nav : {len(df)} rows')

    # Transactions
    df = pd.read_csv(f'{RAW_DIR}/08_investor_transactions.csv')
    df['transaction_date'] = pd.to_datetime(df['transaction_date'], errors='coerce')
    df['amount_inr']       = pd.to_numeric(df['amount_inr'], errors='coerce')
    df = df[df['amount_inr'] > 0]
    df.to_csv(f'{PROCESSED_DIR}/08_transactions_clean.csv', index=False)
    df.to_sql('fact_transactions', engine, if_exists='replace', index=False)
    print(f'fact_transactions : {len(df)} rows')

    print('ETL complete!')
    print(f'Database : {DB_PATH}')

if __name__ == '__main__':
    clean_and_load()
"""

with open("bluestock/scripts/etl_pipeline.py", "w") as f:
    f.write(etl_pipeline)
print("✅ etl_pipeline.py saved")

# ── live_nav_fetch.py ─────────────────────────────────────
live_nav = """
import requests
import pandas as pd
import os

RAW_DIR = 'data/raw'
os.makedirs(RAW_DIR, exist_ok=True)

SCHEMES = {
    125497: 'HDFC_Top100_Direct',
    119551: 'SBI_Bluechip',
    120503: 'ICICI_Bluechip',
    118632: 'Nippon_LargeCap',
    119092: 'Axis_Bluechip',
    120841: 'Kotak_Bluechip',
}

def fetch_all():
    print('LIVE NAV FETCH - mfapi.in')
    all_dfs = []
    for code, name in SCHEMES.items():
        print(f'Fetching {name} ({code})...', end=' ')
        try:
            r    = requests.get(f'https://api.mfapi.in/mf/{code}', timeout=15)
            data = r.json()
            df   = pd.DataFrame(data['data'])
            df['amfi_code']   = code
            df['scheme_name'] = data['meta']['scheme_name']
            df['fund_house']  = data['meta']['fund_house']
            df.to_csv(f'{RAW_DIR}/nav_live_{code}.csv', index=False)
            all_dfs.append(df)
            print(f'Done - {len(df)} records')
        except Exception as e:
            print(f'Error: {e}')
    if all_dfs:
        combined = pd.concat(all_dfs, ignore_index=True)
        combined.to_csv(f'{RAW_DIR}/nav_live_combined.csv', index=False)
        print(f'Combined saved - {len(combined)} total rows')

if __name__ == '__main__':
    fetch_all()
"""

with open("bluestock/scripts/live_nav_fetch.py", "w") as f:
    f.write(live_nav)
print("✅ live_nav_fetch.py saved")

# ── schema.sql ────────────────────────────────────────────
schema_sql = """-- Bluestock Fintech MF Analytics Star Schema
-- Day 2

CREATE TABLE IF NOT EXISTS dim_fund (
    amfi_code           INTEGER PRIMARY KEY,
    fund_house          TEXT,
    scheme_name         TEXT,
    category            TEXT,
    sub_category        TEXT,
    plan                TEXT,
    launch_date         TEXT,
    benchmark           TEXT,
    expense_ratio_pct   REAL,
    exit_load_pct       REAL,
    fund_manager        TEXT,
    risk_category       TEXT
);

CREATE TABLE IF NOT EXISTS dim_investor (
    investor_id         TEXT PRIMARY KEY,
    state               TEXT,
    city                TEXT,
    city_tier           TEXT,
    age_group           TEXT,
    gender              TEXT,
    annual_income_lakh  REAL
);

CREATE TABLE IF NOT EXISTS fact_nav (
    nav_id              INTEGER PRIMARY KEY AUTOINCREMENT,
    amfi_code           INTEGER,
    date                TEXT,
    nav                 REAL,
    daily_return_pct    REAL,
    FOREIGN KEY (amfi_code) REFERENCES dim_fund(amfi_code)
);

CREATE TABLE IF NOT EXISTS fact_transactions (
    txn_id              INTEGER PRIMARY KEY AUTOINCREMENT,
    investor_id         TEXT,
    amfi_code           INTEGER,
    transaction_date    TEXT,
    transaction_type    TEXT,
    amount_inr          INTEGER,
    payment_mode        TEXT,
    kyc_status          TEXT,
    FOREIGN KEY (investor_id) REFERENCES dim_investor(investor_id),
    FOREIGN KEY (amfi_code)   REFERENCES dim_fund(amfi_code)
);

CREATE TABLE IF NOT EXISTS fact_performance (
    perf_id             INTEGER PRIMARY KEY AUTOINCREMENT,
    amfi_code           INTEGER,
    return_1yr_pct      REAL,
    return_3yr_pct      REAL,
    return_5yr_pct      REAL,
    sharpe_ratio        REAL,
    sortino_ratio       REAL,
    beta                REAL,
    alpha               REAL,
    max_drawdown_pct    REAL,
    std_dev_ann_pct     REAL,
    aum_crore           INTEGER,
    FOREIGN KEY (amfi_code) REFERENCES dim_fund(amfi_code)
);

CREATE TABLE IF NOT EXISTS fact_aum (
    aum_id          INTEGER PRIMARY KEY AUTOINCREMENT,
    date            TEXT,
    fund_house      TEXT,
    aum_lakh_crore  REAL,
    aum_crore       INTEGER,
    num_schemes     INTEGER
);

CREATE TABLE IF NOT EXISTS fact_sip (
    sip_id                    INTEGER PRIMARY KEY AUTOINCREMENT,
    month                     TEXT,
    sip_inflow_crore          INTEGER,
    active_sip_accounts_crore REAL,
    new_sip_accounts_lakh     REAL,
    sip_aum_lakh_crore        REAL,
    yoy_growth_pct            REAL
);

-- INDEXES
CREATE INDEX IF NOT EXISTS idx_nav_code  ON fact_nav(amfi_code);
CREATE INDEX IF NOT EXISTS idx_nav_date  ON fact_nav(date);
CREATE INDEX IF NOT EXISTS idx_txn_code  ON fact_transactions(amfi_code);
CREATE INDEX IF NOT EXISTS idx_txn_date  ON fact_transactions(transaction_date);
"""

os.makedirs("bluestock/sql", exist_ok=True)
with open("bluestock/sql/schema.sql", "w") as f:
    f.write(schema_sql)
print("✅ schema.sql saved")

# ── queries.sql ───────────────────────────────────────────
queries_sql = """-- Bluestock Fintech - 10 Analytical SQL Queries

-- Q1: Top 5 funds by AUM
SELECT scheme_name, fund_house, aum_crore
FROM fact_performance
ORDER BY aum_crore DESC LIMIT 5;

-- Q2: Average NAV per month per fund
SELECT amfi_code,
       strftime('%Y-%m', date) AS month,
       ROUND(AVG(nav), 4)      AS avg_nav
FROM fact_nav
GROUP BY amfi_code, month
ORDER BY amfi_code, month;

-- Q3: SIP inflow YoY growth
SELECT month, sip_inflow_crore, yoy_growth_pct
FROM fact_sip
ORDER BY month;

-- Q4: Total transaction amount by state
SELECT state,
       COUNT(*)        AS num_transactions,
       SUM(amount_inr) AS total_amount
FROM fact_transactions ft
JOIN dim_investor di ON ft.investor_id = di.investor_id
GROUP BY state
ORDER BY total_amount DESC;

-- Q5: Funds with expense ratio less than 1 percent
SELECT amfi_code, scheme_name, fund_house, expense_ratio_pct
FROM dim_fund
WHERE expense_ratio_pct < 1.0
ORDER BY expense_ratio_pct;

-- Q6: Top 5 funds by Sharpe Ratio
SELECT scheme_name, fund_house, sharpe_ratio, sortino_ratio
FROM fact_performance
ORDER BY sharpe_ratio DESC LIMIT 5;

-- Q7: Monthly transaction count by type
SELECT strftime('%Y-%m', transaction_date) AS month,
       transaction_type,
       COUNT(*)                             AS count,
       SUM(amount_inr)                     AS total_amount
FROM fact_transactions
GROUP BY month, transaction_type
ORDER BY month;

-- Q8: Benchmark index total return
SELECT index_name,
       MIN(close_value) AS start_value,
       MAX(close_value) AS peak_value,
       ROUND((MAX(close_value) - MIN(close_value))
             / MIN(close_value) * 100, 2) AS total_return_pct
FROM fact_benchmark
GROUP BY index_name;

-- Q9: Sector allocation across all funds
SELECT sector,
       ROUND(SUM(weight_pct), 2)     AS total_weight,
       ROUND(AVG(weight_pct), 2)     AS avg_weight,
       COUNT(DISTINCT amfi_code)     AS num_funds
FROM fact_holdings
GROUP BY sector
ORDER BY total_weight DESC;

-- Q10: SIP amount by age group
SELECT age_group,
       COUNT(*)                  AS num_transactions,
       ROUND(AVG(amount_inr), 2) AS avg_sip_amount,
       SUM(amount_inr)           AS total_invested
FROM fact_transactions ft
JOIN dim_investor di ON ft.investor_id = di.investor_id
WHERE ft.transaction_type = 'Sip'
GROUP BY age_group
ORDER BY avg_sip_amount DESC;
"""

with open("bluestock/sql/queries.sql", "w") as f:
    f.write(queries_sql)
print("✅ queries.sql saved")

print("\n✅ All 4 missing files created!")
```
